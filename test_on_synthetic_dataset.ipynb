#### Loading Class

```python
from MorphoGet import MorphoGet 
```

#### Loading data

```python
from Synthetic_Dataset import df
get = MorphoGet(df)
get.set_landmarks()
```

#### Constructing a shape using averaged points

```python
get.run_gpa()
get.plot_consensus(figsize=(4, 3))
get.KEY_LANDMARKS
```

#### Obtaining flight activity indicators at six landmarks and projection onto the remaining markers.
##### Setting a target points and draw a diagram of the indicator calculation scheme

```python
get.set_fa_params(Basal=1, Apex=5, Costa=3, Tornal=7, TaperCosta=4, TaperMargin=6) 
get.plot_schematic()
```

##### Calculating metrics

```python
get.calculate_fa_metrics(sa_a_comp=0)
get.plot_biplot("sa_biplot", figsize=(4, 3))
get.fa_metrics
```

##### Regression of flight activity indicators onto all landmarks

```python
get.plot_regressions(fa_metrics=["MA_s", "AR_s", "Taper_l", "Taper_K"], invert_sigma=["Taper_l"], n_cols=2, z_ext=1.5)
```

#### Obtaining flight activity indicators at four landmarks

```python
get.set_landmarks(cols=['x1', 'y1'] + ['x3', 'y3'] + ['x5', 'y5'] + ['x7', 'y7'])
get.calculate_fa_metrics(sa_l_comp=0, sa_a_comp=0)
get.plot_regressions(fa_metrics=["MA_l", "AR_l", "SA_l", "SA_a"], n_cols=2, z_ext=3.0, smooth=True, invert_sigma=["SA_l", "SA_a"])
```
